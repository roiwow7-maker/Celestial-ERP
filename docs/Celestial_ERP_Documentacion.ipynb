## 2. Arquitectura general

![Arquitectura general](assets/arquitectura_general.svg)

- Applet: portal y orquestador.
- DATA_scope: remuneraciones, ETL, dashboard, reportes y cargas.
- ERP_api: endpoints JSON iniciales.
- Django Admin: administracion interna.
- SQLite: base local actual.
- Backups: automaticos cada 90 minutos.

## 3. Flujo ETL

![Flujo ETL](assets/flujo_etl.svg)

El ETL recibe Excel historico o CSV transformado, genera `transformed.csv`, separa categorias, crea CSV equivalentes a liquidaciones y puede importar los datos a Django.

## 5. Operacion local

El sistema no esta pensado para internet. El objetivo es uso local o red interna controlada.

Prioridades antes de uso multiusuario real:

1. Login obligatorio.
2. Roles y permisos finos.
3. Estados de empleados.
4. Carga individual controlada.
5. Migracion a PostgreSQL antes de operacion concurrente seria.
